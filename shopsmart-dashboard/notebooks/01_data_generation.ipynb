```python
import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import datetime, timedelta
import os

fake = Faker()
np.random.seed(42)
random.seed(42)

print("Libraries loaded successfully")
```

```python
stores = pd.DataFrame({
    'store_id': ['S001', 'S002', 'S003', 'S004', 'S005'],
    'store_name': ['Downtown', 'Westside', 'Northgate', 'Eastpark', 'Southmall'],
    'city': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'],
    'manager': [fake.name() for _ in range(5)],
    'open_date': pd.to_datetime(['2018-01-15', '2019-03-20', '2017-06-10', '2020-11-05', '2021-02-28'])
})

print(stores)
```

```python
categories = {
    'Electronics': [('Laptop', 899.99), ('Headphones', 79.99), ('Smartwatch', 199.99), ('Tablet', 449.99)],
    'Clothing': [('T-Shirt', 19.99), ('Jeans', 49.99), ('Jacket', 89.99), ('Sneakers', 69.99)],
    'Home': [('Coffee Maker', 59.99), ('Blender', 39.99), ('Vacuum', 129.99), ('Lamp', 34.99)],
    'Food': [('Protein Bar', 2.99), ('Coffee Beans', 14.99), ('Olive Oil', 9.99), ('Granola', 6.99)]
}

products_list = []
product_id = 1

for category, items in categories.items():
    for name, price in items:
        products_list.append({
            'product_id': f'P{product_id:03d}',
            'product_name': name,
            'category': category,
            'price': price,
            'cost': round(price * random.uniform(0.4, 0.6), 2)
        })
        product_id += 1

products = pd.DataFrame(products_list)
print(products)
```

```python
sales_list = []
start_date = datetime(2023, 1, 1)
end_date = datetime(2023, 12, 31)

for _ in range(5000):
    product = products.sample(1).iloc[0]
    store = stores.sample(1).iloc[0]
    date = start_date + timedelta(days=random.randint(0, 364))
    quantity = random.randint(1, 10)

    sales_list.append({
        'sale_id': f'SALE{_ + 1:05d}',
        'date': date.strftime('%Y-%m-%d'),
        'store_id': store['store_id'],
        'store_name': store['store_name'],
        'product_id': product['product_id'],
        'product_name': product['product_name'],
        'category': product['category'],
        'quantity': quantity,
        'unit_price': product['price'],
        'unit_cost': product['cost'],
        'total_revenue': round(product['price'] * quantity, 2),
        'total_cost': round(product['cost'] * quantity, 2),
        'profit': round((product['price'] - product['cost']) * quantity, 2)
    })

sales = pd.DataFrame(sales_list)
print(f"Total sales records: {len(sales)}")
print(sales.head())
```

```python
os.makedirs('../data', exist_ok=True)

stores.to_csv('../data/stores.csv', index=False)
products.to_csv('../data/products.csv', index=False)
sales.to_csv('../data/sales.csv', index=False)

print("Files saved successfully:")
print(f"  - stores.csv: {len(stores)} records")
print(f"  - products.csv: {len(products)} records")
print(f"  - sales.csv: {len(sales)} records")
```
